```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 10000

# Hourly timestamps
timestamps = pd.date_range(
    start="2025-01-01 00:00:00",
    periods=n,
    freq="h"
)

external_context = pd.DataFrame({
    "Timestamp": timestamps,
    "Ambient_Temperature": np.round(np.random.uniform(20, 40, n), 2),
    "Load_Density": np.round(np.random.uniform(30, 100, n), 2),
    "Humidity": np.round(np.random.uniform(40, 90, n), 2)
})

# Shift from timestamp hour
external_context["Shift"] = external_context["Timestamp"].dt.hour.map(
    lambda x:
        "Morning" if 6 <= x < 14 else
        "Evening" if 14 <= x < 22 else
        "Night"
)

# Day Type from actual day
external_context["Day_Type"] = external_context["Timestamp"].dt.dayofweek.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)

print(external_context.shape)
external_context.head()
```
