```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Load timestamps file
df = pd.read_csv(
    r"C:\Users\ankit\Predictive-Maintenance-IoT-AI\Data\timestamps_added.csv"
)

# Create external context dataset
external_context = pd.DataFrame({
    "Timestamp": df["timestamp"],

    "Ambient_Temperature":
        np.random.randint(20, 41, len(df)),

    "Load_Density":
        np.random.randint(30, 101, len(df)),

    "Humidity":
        np.random.randint(40, 91, len(df)),

    "Shift":
        np.random.choice(
            ["Morning", "Evening", "Night"],
            len(df)
        ),

    "Day_Type":
        np.random.choice(
            ["Weekday", "Weekend"],
            len(df)
        )
})

print(external_context.shape)
external_context.head()
```

```python
print(len(df))
print(len(external_context))

print(
    (external_context["Timestamp"] ==
     df["timestamp"]).all()
)
```

```python
external_context.to_csv(
    r"C:\Users\ankit\Predictive-Maintenance-IoT-AI\Data\external_context.csv",
    index=False
)

print("Updated external_context.csv saved")
```
