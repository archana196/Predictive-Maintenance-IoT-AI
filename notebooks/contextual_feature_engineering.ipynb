```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/context_merged_dataset.csv")
print(df.shape)
df.head()
```

```python
print(df.columns)
```

```python
df["temp_diff"] = (
    df["Process temperature [K]"]
    - df["Air temperature [K]"]
)
```

```python
df["temp_diff"].head()
```

```python
df["load_ratio"] = (
    df["Load_Density"] / 100
)
```

```python
df["load_ratio"].head()
```

```python
df["humidity_impact"] = (
    df["Humidity"]
    * df["load_ratio"]
)
```

```python
df["humidity_impact"].head()
```

```python
print(
    df[
        [
            "temp_diff",
            "load_ratio",
            "humidity_impact"
        ]
    ].head()
)
```

```python
print(df.isnull().sum())
```

```python
print(
    df[
        [
            "temp_diff",
            "load_ratio",
            "humidity_impact"
        ]
    ].describe()
)
```

```python
df.to_csv(
    "../data/context_features_dataset.csv",
    index=False
)

print(
    "Saved: context_features_dataset.csv"
)
```
