```python
import pandas as pd
```

```python
df = pd.read_csv("../data/timestamps_added.csv")
context = pd.read_csv("../data/external_context.csv")
```

```python
print(df.shape)
print(context.shape)
```

```python
print(df.columns)
print(context.columns)
```

```python
context.rename(
    columns={"Timestamp": "timestamp"},
    inplace=True
)
```

```python
print(df.columns)
print(context.columns)
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
context["timestamp"] = pd.to_datetime(context["timestamp"])
```

```python
merged_df = pd.merge(
    df,
    context,
    on="timestamp",
    how="left"
)
```

```python
print(merged_df.shape)
merged_df.head()
```

```python
print(merged_df.isnull().sum())
```

```python
merged_df.info()
```

```python
merged_df.to_csv(
    "../data/context_merged_dataset.csv",
    index=False
)

print("Saved: context_merged_dataset.csv")
```
