```python
import pandas as pd

from sklearn.model_selection import StratifiedKFold
```

```python
df = pd.read_csv("../data/contextual_merged_dataset.csv")

print(df.shape)
df.head()
```

```python
target = "Machine failure"

X = df.drop(columns=[target])
y = df[target]

print("Original Class Distribution")
print(y.value_counts())
```

```python
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("Stratified 5-Fold Cross Validation Initialized")
```

```python
fold_results = []

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), start=1):

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    train_counts = y_train.value_counts()
    test_counts = y_test.value_counts()

    fold_results.append([
        fold,
        train_counts.get(0, 0),
        train_counts.get(1, 0),
        test_counts.get(0, 0),
        test_counts.get(1, 0)
    ])

    print(f"\nFold {fold}")
    print("Train:")
    print(train_counts)

    print("Test:")
    print(test_counts)
```

```python
fold_df = pd.DataFrame(
    fold_results,
    columns=[
        "Fold",
        "Train_Class_0",
        "Train_Class_1",
        "Test_Class_0",
        "Test_Class_1"
    ]
)

fold_df
```

```python
print("\nAll folds maintain similar class distribution.")
print("Stratified K-Fold verification completed.")
```
