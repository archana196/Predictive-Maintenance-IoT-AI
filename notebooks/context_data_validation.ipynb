```python
import pandas as pd
```

```python
telemetry = pd.read_csv(
    r"C:\Users\ankit\Predictive-Maintenance-IoT-AI\Data\timestamps_added.csv"
)

context = pd.read_csv(
    r"C:\Users\ankit\Predictive-Maintenance-IoT-AI\Data\external_context.csv"
)

print(telemetry.shape)
print(context.shape)
```

```python
merged_df = pd.merge(
    telemetry,
    context,
    left_on="timestamp",
    right_on="Timestamp",
    how="left"
)

print(merged_df.shape)
```

```python
print("Rows:", len(merged_df))
print("Columns:", len(merged_df.columns))

print("Missing Values:")
print(merged_df.isnull().sum().sum())

print("Duplicate Records:")
print(merged_df.duplicated().sum())
```

```python
print(merged_df.dtypes)
```

```python
merged_df.to_csv(
    r"C:\Users\ankit\Predictive-Maintenance-IoT-AI\Data\contextual_merged_dataset.csv",
    index=False
)

print("Dataset saved successfully")
```
