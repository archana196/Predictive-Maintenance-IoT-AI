```python
import pandas as pd
```

```python
df = pd.read_csv("../data/ai4i2020.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df['timestamp'] = pd.date_range(
    start='2025-01-01 00:00:00',
    periods=len(df),
    freq='1min'
)
```

```python
df[['timestamp']].head()
```

```python
df[['timestamp']].tail()
```

```python
df['timestamp'].dtype
```

```python
df.to_csv(
    "../data/timestamps_added.csv",
    index=False
)

print("Saved: timestamps_added.csv")
```

```python
context = pd.DataFrame({
    "timestamp": df["timestamp"],
    "ambient_temperature": None,
    "load_density": None,
    "humidity": None,
    "shift": None,
    "day_type": None
})

context.head()
```

```python
context.to_csv(
    "../data/context_structure.csv",
    index=False
)

print("Saved: context_structure.csv")
```
