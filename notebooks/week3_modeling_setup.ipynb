```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/context_features_dataset.csv")

print(df.shape)
df.head()
```

```python
print(df.columns)
```

```python
df.drop(
    columns=["UDI", "Product ID"],
    inplace=True,
    errors="ignore"
)

print(df.shape)
```

```python
print(df.isnull().sum().sum())
```

```python
print(df.dtypes)
```

```python
categorical_cols = ["Type", "Shift", "Day_Type"]

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype("category").cat.codes
```

```python
print(df.head())
```

```python
print(df["Machine failure"].value_counts())
```

```python
failure_rate = (
    df["Machine failure"].mean() * 100
)

print(
    f"Failure Rate: {failure_rate:.2f}%"
)
```

```python
X = df.drop(
    columns=["Machine failure"]
)

y = df["Machine failure"]

print(X.shape)
print(y.shape)
```

```python
print(y.value_counts())
```

```python
print(
    "Dataset prepared for Stratified K-Fold and SMOTE."
)
```

```python
prepared_df = pd.concat(
    [X, y],
    axis=1
)

prepared_df.to_csv(
    "../data/model_ready_dataset.csv",
    index=False
)

print("Saved model_ready_dataset.csv")
```
