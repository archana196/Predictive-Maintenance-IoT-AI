```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/week1_features.csv")
print("Shape:", df.shape)
df.head()
```

```python
df["temp_difference"] = df["Process temperature [K]"] - df["Air temperature [K]"]
print("Temp difference added!")
```

```python
df["load_ratio"] = df["Torque [Nm]"] / (df["Rotational speed [rpm]"] + 1)
print("Load ratio added!")
```

```python
df["shift"] = (df.index % 24 // 8)
print("Shift encoding added!")
```

```python
df["temp_torque"] = df["Air temperature [K]"] * df["Torque [Nm]"]
df["rpm_wear"] = df["Rotational speed [rpm]"] * df["Tool wear [min]"]
print("Interaction features added!")
print("Total features:", df.shape[1])
```
